```python
import requests
import sqlite3
import pandas as pd
from bs4 import BeautifulSoup

url = 'https://web.archive.org/web/20230902185655/https://en.everybodywiki.com/100_Most_Highly-Ranked_Films'
db_name = 'Movies.db'   #Nom de la base de données
table_name = 'Top_50'   #Nom de la table
csv_path = 'top_50_films.csv'  #Nom du fichier csv auquel les données vont être déposé avant d'être acheminé dans la table de la base de données

df = pd.DataFrame(columns=["Average Rank","Film","Year"])  #Création d'un dataframe vide
count = 0    #Initialisation pour boucle

#Ce script permet d'avoir accés aux différentes tables de la page web
html_page = requests.get(url).text
data = BeautifulSoup(html_page, 'html.parser')
tables = data.find_all('tbody')

#Ce script permet de récupérer toutes les lignes de la première table
rows = tables[0].find_all('tr')

#Ce boucle permet de récupérer les 50 premières ligne de la première table, extraire les données et les insérées dans un dataframe
for row in rows:
    if count < 50 :
        col = row.find_all('td')
        if len(col)!= 0:
            data_dict = {"Average Rank": int(col[0].contents[0]),
                         "Film": str(col[1].contents[0]),
                         "Year": int(col[2].contents[0])}
            df1 = pd.DataFrame(data_dict, index = [0])
            df = pd.concat([df,df1], ignore_index = True)
            count += 1
    else:
        break

print(df)   #Affiche le dataframe de sorti

df.to_csv(csv_path)    #Charge les données du dataframe de sorti dans un fichier csv

#Ce script permet de créer une connexion avec sqlite3 et de charger le fichier csv dans la table de la base de données
conn = sqlite3.connect(db_name)

df.to_sql(table_name, conn, if_exists = 'replace', index = False)
conn.close()
```
