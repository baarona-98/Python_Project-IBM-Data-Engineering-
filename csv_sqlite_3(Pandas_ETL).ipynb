```python
import requests
import sqlite3
import pandas as pd

url = (
    "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/"
    "IBMSkillsNetwork-PY0221EN-Coursera/labs/v2/INSTRUCTOR.csv"
)
r = requests.get(url)
with open ("INSTRUCTOR.csv", 'wb') as f:
    f.write (r.content)
```

```python
# Connexion et création de la base de données sur sqlite3
conn = sqlite3.connect('STAFF.db')

# Paramètres de la table
table_name = 'INSTRUCTOR'
attribute_list = ['ID', 'FNAME', 'LNAME', 'CITY', 'CCODE']

# Lecture du fichier csv
df = pd.read_csv("INSTRUCTOR.csv", names = attribute_list)

# Chargement des données sous format csv
df.to_sql(table_name, conn, if_exists = 'replace', index = False)
print('Table is ready')

# Requêtage sql
query_statement = f"SELECT * FROM {table_name}"
query_output = pd.read_sql(query_statement, conn)
print(query_statement)
print(query_output)

query_statement = f"SELECT FNAME FROM {table_name}"
query_output = pd.read_sql(query_statement, conn)
print(query_statement)
print(query_output)

query_statement = f"SELECT COUNT(*) FROM {table_name}"
query_output = pd.read_sql(query_statement, conn)
print(query_statement)
print(query_output)

# Ce script crée un dictionnaire et le transforme en dataframe
data_dict = {'ID' : [100],
            'FNAME' : ['John'],
            'LNAME' : ['Doe'],
            'CITY' : ['Paris'],
            'CCODE' : ['FR']}
data_append = pd.DataFrame(data_dict)

# Ajout de données dans la table
data_append.to_sql(table_name, conn, if_exists = 'append', index = False)
print('Data appended successfully')

# Requête sql
query_statement = f"SELECT COUNT(*) FROM {table_name}"
query_output = pd.read_sql(query_statement, conn)
print(query_statement)
print(query_output)

# Fermeture de la connexion
conn.close()
```
